```python
# ============================================================
# RESNET-STYLE CNN ON MNIST
# ============================================================

import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import layers, Model
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------

EPOCHS = 10
BATCH_SIZE = 128
DROPOUT_RATE = 0.20

# ------------------------------------------------------------
# Load MNIST
# ------------------------------------------------------------

(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Add channel dimension
x_train = x_train[..., tf.newaxis]
x_test = x_test[..., tf.newaxis]

# Normalize
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# One-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print("Training data:", x_train.shape)
print("Testing data :", x_test.shape)


# ------------------------------------------------------------
# Residual Block
# ------------------------------------------------------------

def residual_block(x, filters, dropout_rate=0.2):

    shortcut = x

    # First convolution
    y = layers.Conv2D(
        filters,
        kernel_size=3,
        padding="same",
        activation="relu"
    )(x)

    y = layers.BatchNormalization()(y)

    # Second convolution
    y = layers.Conv2D(
        filters,
        kernel_size=3,
        padding="same"
    )(y)

    y = layers.BatchNormalization()(y)

    # Match dimensions for shortcut
    if x.shape[-1] != filters:
        shortcut = layers.Conv2D(
            filters,
            kernel_size=1,
            padding="same"
        )(shortcut)

        shortcut = layers.BatchNormalization()(shortcut)

    # Residual connection
    y = layers.Add()([shortcut, y])

    # Activation
    y = layers.Activation("relu")(y)

    # Dropout
    y = layers.Dropout(dropout_rate)(y)

    return y


# ------------------------------------------------------------
# Build ResNet
# ------------------------------------------------------------

inputs = layers.Input(shape=(28, 28, 1))

# Initial convolution
x = layers.Conv2D(
    32,
    kernel_size=3,
    padding="same",
    activation="relu"
)(inputs)

x = layers.BatchNormalization()(x)


# ---------------- Residual Block 1 ----------------

x = residual_block(
    x,
    filters=32,
    dropout_rate=DROPOUT_RATE
)

x = layers.MaxPooling2D(pool_size=2)(x)


# ---------------- Residual Block 2 ----------------

x = residual_block(
    x,
    filters=64,
    dropout_rate=DROPOUT_RATE
)

x = layers.MaxPooling2D(pool_size=2)(x)


# ---------------- Residual Block 3 ----------------

x = residual_block(
    x,
    filters=128,
    dropout_rate=DROPOUT_RATE
)

x = layers.MaxPooling2D(pool_size=2)(x)


# ---------------- Residual Block 4 ----------------

x = residual_block(
    x,
    filters=256,
    dropout_rate=DROPOUT_RATE
)


# ------------------------------------------------------------
# Classification
# ------------------------------------------------------------

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dropout(DROPOUT_RATE)(x)

outputs = layers.Dense(
    10,
    activation="softmax"
)(x)


# ------------------------------------------------------------
# Create Model
# ------------------------------------------------------------

model = Model(inputs, outputs)

model.summary()


# ------------------------------------------------------------
# Compile
# ------------------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

history = model.fit(
    x_train,
    y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.10,
    verbose=1
)


# ------------------------------------------------------------
# Test
# ------------------------------------------------------------

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nTest Loss:", test_loss)
print("Test Accuracy:", test_accuracy)


# ------------------------------------------------------------
# Plot Accuracy
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("ResNet on MNIST")

plt.legend()
plt.grid(True)

plt.show()


```

```python
# ============================================================
# DENSENET-STYLE CNN ON MNIST
# ============================================================

import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import layers, Model
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------

EPOCHS = 10
BATCH_SIZE = 128

DROPOUT_RATE = 0.20
GROWTH_RATE = 16


# ------------------------------------------------------------
# Load MNIST
# ------------------------------------------------------------

(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train[..., tf.newaxis]
x_test = x_test[..., tf.newaxis]

# Normalize
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# One-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print("Training data:", x_train.shape)
print("Testing data :", x_test.shape)


# ------------------------------------------------------------
# Dense Layer
# ------------------------------------------------------------

def dense_layer(x):

    # Batch Normalization
    y = layers.BatchNormalization()(x)

    # ReLU
    y = layers.Activation("relu")(y)

    # Convolution
    y = layers.Conv2D(
        GROWTH_RATE,
        kernel_size=3,
        padding="same"
    )(y)

    # Dropout
    y = layers.Dropout(
        DROPOUT_RATE
    )(y)

    # Dense connection
    y = layers.Concatenate()([x, y])

    return y


# ------------------------------------------------------------
# Dense Block
# ------------------------------------------------------------

def dense_block(x, number_of_layers):

    for i in range(number_of_layers):

        x = dense_layer(x)

    return x


# ------------------------------------------------------------
# Transition Block
# ------------------------------------------------------------

def transition_block(x):

    # Batch Normalization
    x = layers.BatchNormalization()(x)

    # ReLU
    x = layers.Activation("relu")(x)

    # 1x1 convolution
    filters = int(x.shape[-1] * 0.5)

    x = layers.Conv2D(
        filters,
        kernel_size=1,
        padding="same",
        activation="relu"
    )(x)

    x = layers.Dropout(
        DROPOUT_RATE
    )(x)

    # Max pooling
    x = layers.MaxPooling2D(
        pool_size=2
    )(x)

    return x


# ------------------------------------------------------------
# Build DenseNet
# ------------------------------------------------------------

inputs = layers.Input(
    shape=(28, 28, 1)
)

# Initial convolution
x = layers.Conv2D(
    32,
    kernel_size=3,
    padding="same",
    activation="relu"
)(inputs)


# ================= Dense Block 1 =================

x = dense_block(
    x,
    number_of_layers=2
)

x = transition_block(x)


# ================= Dense Block 2 =================

x = dense_block(
    x,
    number_of_layers=2
)

x = transition_block(x)


# ================= Dense Block 3 =================

x = dense_block(
    x,
    number_of_layers=2
)

x = transition_block(x)


# ================= Dense Block 4 =================

x = dense_block(
    x,
    number_of_layers=2
)


# ------------------------------------------------------------
# Classification
# ------------------------------------------------------------

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dropout(
    DROPOUT_RATE
)(x)

outputs = layers.Dense(
    10,
    activation="softmax"
)(x)


# ------------------------------------------------------------
# Create Model
# ------------------------------------------------------------

model = Model(
    inputs,
    outputs
)

model.summary()


# ------------------------------------------------------------
# Compile
# ------------------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

history = model.fit(
    x_train,
    y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.10,
    verbose=1
)


# ------------------------------------------------------------
# Test
# ------------------------------------------------------------

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nTest Loss:", test_loss)
print("Test Accuracy:", test_accuracy)


# ------------------------------------------------------------
# Plot Accuracy
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("DenseNet on MNIST")

plt.legend()
plt.grid(True)

plt.show()

```

### 1. Preparing the Image Super-Resolution Dataset (MNIST Downsampling)
We will downsample the $28 \times 28$ MNIST images to $14 \times 14$ using average pooling to create our low-resolution (LR) training inputs. The original $28 \times 28$ images will serve as our high-resolution (HR) targets.

```python
import numpy as np

# Downsample 28x28 images to 14x14 using average pooling
x_train_lr = tf.keras.layers.AveragePooling2D(pool_size=2)(x_train).numpy()
x_test_lr = tf.keras.layers.AveragePooling2D(pool_size=2)(x_test).numpy()

print("Low-Resolution train shape:", x_train_lr.shape)
print("High-Resolution train shape:", x_train.shape)
```

### 2. Designing the Residual Dense Block (RDB) & RDN Autoencoder
We construct a Residual Dense Block where features from all preceding convolutional layers are concatenated, and a local residual connection adds the block's input back to the aggregated features.
We then build an Autoencoder that takes the $14 \times 14 \times 1$ LR images, processes them through RDBs, and upscales them back to $28 \times 28 \times 1$ using sub-pixel convolution (`Conv2DTranspose` or `UpSampling2D`).

```python
def residual_dense_block(x, filters=32, growth_rate=16, num_layers=3):
    """
    A Residual Dense Block (RDB) combining dense connections with a local residual connection.
    """
    local_features = x
    concat_features = x

    for i in range(num_layers):
        # Conv -> BN -> Activation
        y = layers.Conv2D(growth_rate, kernel_size=3, padding="same")(concat_features)
        y = layers.BatchNormalization()(y)
        y = layers.Activation("relu")(y)

        # Concatenate preceding features (Dense connection)
        concat_features = layers.Concatenate()([concat_features, y])

    # 1x1 convolution to match local features channel dimension
    local_residual = layers.Conv2D(filters, kernel_size=1, padding="same")(concat_features)

    # Local Residual Connection
    output = layers.Add()([local_features, local_residual])
    return output

def build_rdn_super_resolution_model(input_shape=(14, 14, 1)):
    inputs = layers.Input(shape=input_shape)

    # Initial Shallow Feature Extraction
    x = layers.Conv2D(64, kernel_size=3, padding="same", activation="relu")(inputs)
    x1 = layers.BatchNormalization()(x)

    # Residual Dense Blocks (RDBs)
    x = residual_dense_block(x1, filters=64, growth_rate=16, num_layers=3)
    x = residual_dense_block(x, filters=64, growth_rate=16, num_layers=3)

    # Global Residual Connection
    x = layers.Add()([x1, x])

    # Upscaling (Super-Resolution Reconstruction from 14x14 to 28x28)
    x = layers.Conv2DTranspose(32, kernel_size=3, strides=2, padding="same", activation="relu")(x)
    x = layers.BatchNormalization()(x)

    # Final reconstruction layer
    outputs = layers.Conv2D(1, kernel_size=3, padding="same", activation="sigmoid")(x)

    model = Model(inputs, outputs, name="RDN_SuperResolution")
    return model

sr_model = build_rdn_super_resolution_model()
sr_model.summary()
```

### 3. Training the Super-Resolution Model
We compile the RDN autoencoder using Mean Squared Error (MSE) loss, which is standard for image restoration and super-resolution tasks.

```python
sr_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mean_squared_error",
    metrics=["mae"]
)

# Train using low-resolution inputs and original high-resolution inputs as targets
history_sr = sr_model.fit(
    x_train_lr,
    x_train,
    epochs=5,
    batch_size=128,
    validation_split=0.10,
    verbose=1
)
```

### 4. Visualizing Super-Resolution Performance
Let's plot side-by-side comparisons of the Low-Resolution Input, the Ground Truth High-Resolution image, and our model's reconstructed Super-Resolved output.

```python
# Predict on test set
decoded_imgs = sr_model.predict(x_test_lr)

n = 8
plt.figure(figsize=(15, 6))
for i in range(n):
    # 1. Low-Resolution
    ax = plt.subplot(3, n, i + 1)
    plt.imshow(x_test_lr[i].reshape(14, 14), cmap="gray")
    plt.title("Low-Res")
    plt.axis("off")

    # 2. Super-Resolved Reconstruction
    ax = plt.subplot(3, n, i + 1 + n)
    plt.imshow(decoded_imgs[i].reshape(28, 28), cmap="gray")
    plt.title("Super-Resolved")
    plt.axis("off")

    # 3. Ground Truth Original
    ax = plt.subplot(3, n, i + 1 + 2 * n)
    plt.imshow(x_test[i].reshape(28, 28), cmap="gray")
    plt.title("Ground Truth")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
