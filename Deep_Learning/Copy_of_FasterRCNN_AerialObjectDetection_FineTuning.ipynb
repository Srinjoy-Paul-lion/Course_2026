```python
# ================================================================
# FASTER R-CNN FOR AERIAL OBJECT DETECTION
# Kaggle dataset:
# simplestzyp/tiny-object-detection-in-aerial-images
#
# Complete Google Colab pipeline:
# 1. Configure Kaggle API
# 2. Download and extract dataset
# 3. Find COCO JSON annotations
# 4. Load images and bounding boxes
# 5. Split training and validation data if needed
# 6. Fine-tune Faster R-CNN ResNet50 FPN
# 7. Visualize predictions
# 8. Save trained model
# ================================================================

import os
import sys
import json
import time
import random
import shutil
import subprocess
from pathlib import Path
from collections import defaultdict

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageDraw

import torch
import torchvision

from google.colab import files

from torch.utils.data import Dataset, DataLoader
from torchvision.transforms.functional import to_tensor

from torchvision.models.detection import (
    fasterrcnn_resnet50_fpn,
    FasterRCNN_ResNet50_FPN_Weights
)

from torchvision.models.detection.faster_rcnn import FastRCNNPredictor
from torchvision.models.detection.anchor_utils import AnchorGenerator


# ================================================================
# 1. CONFIGURATION
# ================================================================

SEED = 42
NUM_EPOCHS = 5
BATCH_SIZE = 2
CONFIDENCE_THRESHOLD = 0.30

DATASET_ID = "simplestzyp/tiny-object-detection-in-aerial-images"

DATASET_DIR = Path("/content/aerial_dataset")
CHECKPOINT_PATH = "/content/fasterrcnn_aerial_final.pth"

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 65)
print("FASTER R-CNN AERIAL OBJECT DETECTION")
print("=" * 65)

print("PyTorch:", torch.__version__)
print("Torchvision:", torchvision.__version__)
print("Device:", device)

if device.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))


# ================================================================
# 2. INSTALL KAGGLE API AND UPLOAD CREDENTIALS
# ================================================================

subprocess.check_call([
    sys.executable,
    "-m",
    "pip",
    "install",
    "-q",
    "kaggle"
])

print("\nUpload your Kaggle API token: kaggle.json")

uploaded = files.upload()

if "kaggle.json" not in uploaded:
    raise FileNotFoundError(
        "Please upload your Kaggle API credentials named kaggle.json."
    )

os.makedirs("/root/.kaggle", exist_ok=True)

shutil.copy(
    "/content/kaggle.json",
    "/root/.kaggle/kaggle.json"
)

os.chmod(
    "/root/.kaggle/kaggle.json",
    0o600
)

os.remove("/content/kaggle.json")

print("Kaggle API configured successfully.")


# ================================================================
# 3. DOWNLOAD DATASET
# ================================================================

DATASET_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("\nDownloading Kaggle dataset...")

subprocess.run(
    [
        "kaggle",
        "datasets",
        "download",
        "-d",
        DATASET_ID,
        "-p",
        str(DATASET_DIR),
        "--unzip"
    ],
    check=True
)

print("Dataset downloaded:", DATASET_DIR)


# ================================================================
# 4. FIND IMAGES AND JSON FILES
# ================================================================

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp",
    ".tif", ".tiff", ".webp"
}

all_files = [
    p for p in DATASET_DIR.rglob("*")
    if p.is_file()
]

image_files = [
    p for p in all_files
    if p.suffix.lower() in IMAGE_EXTENSIONS
]

json_files = [
    p for p in all_files
    if p.suffix.lower() == ".json"
]

print("\nTotal files:", len(all_files))
print("Image files:", len(image_files))
print("JSON files:", len(json_files))

print("\nJSON files found:")

for path in json_files:
    print(path.relative_to(DATASET_DIR))

if not image_files:
    raise RuntimeError(
        "No images found in the downloaded dataset."
    )


# ================================================================
# 5. IDENTIFY COCO-STYLE ANNOTATION FILES
# ================================================================

coco_candidates = []

for json_path in json_files:

    try:
        with open(json_path, "r") as f:
            data = json.load(f)

        if (
            isinstance(data, dict)
            and isinstance(data.get("images"), list)
            and isinstance(data.get("annotations"), list)
            and isinstance(data.get("categories"), list)
        ):
            coco_candidates.append({
                "path": json_path,
                "data": data
            })

    except (json.JSONDecodeError, OSError):
        continue

print("\nCOCO annotation candidates:")

for item in coco_candidates:
    data = item["data"]

    print(
        item["path"].name,
        "| Images:", len(data["images"]),
        "| Annotations:", len(data["annotations"]),
        "| Categories:", len(data["categories"])
    )

if not coco_candidates:
    raise RuntimeError(
        "No COCO-style JSON annotation file was found. "
        "The dataset may use a different annotation format. "
        "Inspect the JSON files and adapt the parser before training."
    )


# ================================================================
# 6. SELECT TRAINING AND VALIDATION JSON FILES
# ================================================================

def select_split(items, split):

    candidates = []

    for item in items:

        name = item["path"].stem.lower()
        tokens = name.replace("-", "_").split("_")

        if split == "train":

            if "train" not in tokens:
                continue

            if any(
                token in tokens
                for token in ["val", "valid", "validation", "test"]
            ):
                continue

        elif split == "val":

            if not any(
                token in tokens
                for token in ["val", "valid", "validation"]
            ):
                continue

            if "test" in tokens or "train" in tokens:
                continue

        candidates.append(item)

    if not candidates:
        return None

    candidates.sort(
        key=lambda item: len(item["data"]["images"]),
        reverse=True
    )

    return candidates[0]


train_item = select_split(coco_candidates, "train")
val_item = select_split(coco_candidates, "val")

if train_item is not None and val_item is not None:

    train_json_path = train_item["path"]
    train_json = train_item["data"]

    val_json_path = val_item["path"]
    val_json = val_item["data"]

    separate_splits = True

    print("\nTraining annotations:", train_json_path)
    print("Validation annotations:", val_json_path)

else:

    # Use a single annotation file and create a random split.
    candidates = [
        item for item in coco_candidates
        if "test" not in item["path"].stem.lower().split("_")
    ]

    if not candidates:
        candidates = coco_candidates

    candidates.sort(
        key=lambda item: len(item["data"]["images"]),
        reverse=True
    )

    train_item = candidates[0]

    train_json_path = train_item["path"]
    train_json = train_item["data"]

    val_json_path = None
    val_json = None

    separate_splits = False

    print("\nUsing one annotation file:", train_json_path)
    print("Creating training and validation splits.")


# ================================================================
# 7. INDEX IMAGE PATHS
# ================================================================

image_index = defaultdict(list)

for image_path in image_files:
    image_index[image_path.name].append(image_path)


def resolve_image_path(file_name, annotation_path):

    file_name = str(file_name).replace("\\", "/")
    relative_path = Path(file_name)

    candidates = [
        DATASET_DIR / relative_path,
        annotation_path.parent / relative_path
    ]

    for candidate in candidates:
        if candidate.is_file():
            return candidate

    matches = image_index.get(relative_path.name, [])

    if len(matches) == 1:
        return matches[0]

    if len(matches) > 1:
        raise RuntimeError(
            f"Duplicate filename {relative_path.name}. "
            "Specify its full path in the dataset."
        )

    return None


# ================================================================
# 8. MAP CATEGORY IDS TO CONTIGUOUS LABELS
# ================================================================

categories = sorted(
    train_json["categories"],
    key=lambda x: x["id"]
)

category_id_to_label = {
    category["id"]: index + 1
    for index, category in enumerate(categories)
}

class_names = [
    category["name"]
    for category in categories
]

category_name_to_label = {
    category["name"]: category_id_to_label[category["id"]]
    for category in categories
}

num_classes = len(class_names) + 1

print("\nObject classes:", class_names)
print("Number of foreground classes:", len(class_names))
print("Number of model classes including background:", num_classes)


# ================================================================
# 9. PARSE JSON INTO IMAGE RECORDS
# ================================================================

def parse_coco_records(
    annotation_data,
    annotation_path,
    category_name_to_label
):

    annotations_by_image = defaultdict(list)

    for ann in annotation_data["annotations"]:
        annotations_by_image[ann["image_id"]].append(ann)

    local_categories = {
        cat["id"]: cat["name"]
        for cat in annotation_data["categories"]
    }

    local_category_to_label = {
        cat_id: category_name_to_label[name]
        for cat_id, name in local_categories.items()
        if name in category_name_to_label
    }

    records = []
    missing = 0
    invalid = 0

    for image_info in annotation_data["images"]:

        image_path = resolve_image_path(
            image_info["file_name"],
            annotation_path
        )

        if image_path is None:
            missing += 1
            continue

        with Image.open(image_path) as image:
            width, height = image.size

        boxes = []
        labels = []

        for ann in annotations_by_image[image_info["id"]]:

            if ann.get("iscrowd", 0) == 1:
                continue

            category_id = ann.get("category_id")

            if category_id not in local_category_to_label:
                continue

            bbox = ann.get("bbox")

            if bbox is None or len(bbox) != 4:
                invalid += 1
                continue

            x, y, w, h = map(float, bbox)

            # COCO [x, y, w, h] -> Faster R-CNN [x1, y1, x2, y2]
            x1 = max(0.0, x)
            y1 = max(0.0, y)
            x2 = min(float(width), x + w)
            y2 = min(float(height), y + h)

            if x2 <= x1 or y2 <= y1:
                invalid += 1
                continue

            boxes.append([x1, y1, x2, y2])
            labels.append(
                local_category_to_label[category_id]
            )

        records.append({
            "image_path": str(image_path),
            "image_id": image_info["id"],
            "boxes": boxes,
            "labels": labels
        })

    print("\nParsed:", annotation_path.name)
    print("Usable images:", len(records))
    print("Missing images:", missing)
    print("Invalid boxes:", invalid)

    return records


if separate_splits:

    train_records = parse_coco_records(
        train_json,
        train_json_path,
        category_name_to_label
    )

    val_records = parse_coco_records(
        val_json,
        val_json_path,
        category_name_to_label
    )

else:

    all_records = parse_coco_records(
        train_json,
        train_json_path,
        category_name_to_label
    )

    random.seed(SEED)
    random.shuffle(all_records)

    if len(all_records) < 2:
        raise RuntimeError(
            "At least two images are required for training and validation."
        )

    split_index = int(0.8 * len(all_records))
    split_index = max(
        1,
        min(split_index, len(all_records) - 1)
    )

    train_records = all_records[:split_index]
    val_records = all_records[split_index:]


if not train_records or not val_records:
    raise RuntimeError(
        "Training or validation dataset is empty. "
        "Check the annotation paths and image filenames."
    )

if not any(r["boxes"] for r in train_records):
    raise RuntimeError(
        "No valid training boxes were found."
    )

print("\nTraining images:", len(train_records))
print("Validation images:", len(val_records))


# ================================================================
# 10. VISUALIZE A GROUND-TRUTH IMAGE
# ================================================================

sample_record = next(
    r for r in train_records
    if len(r["boxes"]) > 0
)

sample_image = Image.open(
    sample_record["image_path"]
).convert("RGB")

draw = ImageDraw.Draw(sample_image)

for box, label in zip(
    sample_record["boxes"],
    sample_record["labels"]
):

    x1, y1, x2, y2 = box

    draw.rectangle(
        [x1, y1, x2, y2],
        outline="lime",
        width=2
    )

    draw.text(
        (x1, max(0, y1 - 15)),
        class_names[label - 1],
        fill="yellow"
    )

plt.figure(figsize=(10, 10))
plt.imshow(sample_image)
plt.axis("off")
plt.title("Ground Truth Annotations")
plt.show()


# ================================================================
# 11. PYTORCH DATASET
# ================================================================

class AerialDataset(Dataset):

    def __init__(self, records):
        self.records = records

    def __len__(self):
        return len(self.records)

    def __getitem__(self, index):

        record = self.records[index]

        image = Image.open(
            record["image_path"]
        ).convert("RGB")

        image_tensor = to_tensor(image)

        boxes = torch.tensor(
            record["boxes"],
            dtype=torch.float32
        ).reshape(-1, 4)

        labels = torch.tensor(
            record["labels"],
            dtype=torch.int64
        )

        target = {
            "boxes": boxes,
            "labels": labels,
            "image_id": torch.tensor(
                record["image_id"],
                dtype=torch.int64
            ),
            "area": (
                (boxes[:, 2] - boxes[:, 0]) *
                (boxes[:, 3] - boxes[:, 1])
            ),
            "iscrowd": torch.zeros(
                len(labels),
                dtype=torch.int64
            )
        }

        return image_tensor, target


def collate_fn(batch):
    return tuple(zip(*batch))


train_dataset = AerialDataset(train_records)
val_dataset = AerialDataset(val_records)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    collate_fn=collate_fn,
    pin_memory=torch.cuda.is_available()
)

val_loader = DataLoader(
    val_dataset,
    batch_size=1,
    shuffle=False,
    num_workers=2,
    collate_fn=collate_fn,
    pin_memory=torch.cuda.is_available()
)


# ================================================================
# 12. INITIALIZE FASTER R-CNN
# ================================================================

# Smaller anchors are an experimental starting point for tiny objects.
anchor_generator = AnchorGenerator(
    sizes=((8,), (16,), (32,), (64,), (128,)),
    aspect_ratios=(
        (0.5, 1.0, 2.0),
        (0.5, 1.0, 2.0),
        (0.5, 1.0, 2.0),
        (0.5, 1.0, 2.0),
        (0.5, 1.0, 2.0)
    )
)

####Write your own code to load Faster R-CNN weights
# Start Here




#End here.


# ================================================================
# 13. FREEZE EARLY LAYERS AND FINE-TUNE LATER LAYERS
# ================================================================

for param in model.backbone.parameters():
    param.requires_grad = False

for param in model.backbone.body.layer3.parameters():
    param.requires_grad = True

for param in model.backbone.body.layer4.parameters():
    param.requires_grad = True

for param in model.backbone.fpn.parameters():
    param.requires_grad = True

for param in model.rpn.parameters():
    param.requires_grad = True

for param in model.roi_heads.parameters():
    param.requires_grad = True

print("\nTrainable parameters:", sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
))


# ================================================================
# 14. TRAINING CONFIGURATION
# ================================================================

backbone_params = [
    p for p in model.backbone.parameters()
    if p.requires_grad
]

detection_params = [
    p for name, p in model.named_parameters()
    if p.requires_grad and not name.startswith("backbone.")
]

optimizer = torch.optim.SGD(
    [
        {
            "params": backbone_params,
            "lr": 0.0001
        },
        {
            "params": detection_params,
            "lr": 0.001
        }
    ],
    momentum=0.9,
    weight_decay=0.0005
)

scheduler = torch.optim.lr_scheduler.StepLR(
    optimizer,
    step_size=3,
    gamma=0.1
)

CHECKPOINT_DIR = "/content/aerial_checkpoints"
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

history = []


# ================================================================
# 15. TRAIN FASTER R-CNN
# ================================================================

print("\nStarting training...")

for epoch in range(NUM_EPOCHS):

    #Write your own code to train model.
    # Start Here


    # End Here

    # Keep frozen backbone BatchNorm layers in evaluation mode.
    model.backbone.body.eval()

    model.backbone.body.layer3.train()
    model.backbone.body.layer4.train()

    model.backbone.fpn.train()
    model.rpn.train()
    model.roi_heads.train()

    running_loss = 0.0
    start_time = time.perf_counter()

    for batch_idx, (images, targets) in enumerate(train_loader):

        images = [
            image.to(device)
            for image in images
        ]

        targets = [
            {
                key: value.to(device)
                for key, value in target.items()
            }
            for target in targets
        ]

        optimizer.zero_grad(set_to_none=True)

        loss_dict = model(images, targets)

        total_loss = sum(loss for loss in loss_dict.values())

        if not torch.isfinite(total_loss):
            raise RuntimeError(
                f"Non-finite loss: {total_loss.item()}"
            )

        total_loss.backward()

        torch.nn.utils.clip_grad_norm_(
            [
                p for p in model.parameters()
                if p.requires_grad
            ],
            max_norm=5.0
        )

        optimizer.step()

        running_loss += total_loss.item()

        if (batch_idx + 1) % 50 == 0:
            print(
                f"Epoch {epoch + 1}/{NUM_EPOCHS} "
                f"| Batch {batch_idx + 1}/{len(train_loader)} "
                f"| Loss {total_loss.item():.4f}"
            )

    scheduler.step()

    average_loss = running_loss / max(
        len(train_loader), 1
    )

    history.append(average_loss)

    print(
        f"Epoch {epoch + 1}/{NUM_EPOCHS} "
        f"| Average loss: {average_loss:.4f} "
        f"| Time: {time.perf_counter() - start_time:.1f}s"
    )

    torch.save(
        {
            "epoch": epoch + 1,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "class_names": class_names,
            "num_classes": num_classes,
            "history": history
        },
        os.path.join(
            CHECKPOINT_DIR,
            f"epoch_{epoch + 1}.pth"
        )
    )

print("\nTraining complete.")


# ================================================================
# 16. PLOT TRAINING LOSS
# ================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, len(history) + 1),
    history,
    marker="o"
)

plt.xlabel("Epoch")
plt.ylabel("Average training loss")
plt.title("Faster R-CNN Training Loss")
plt.grid(True)
plt.show()


# ================================================================
# 17. OBJECT DETECTION AND VISUALIZATION
# ================================================================

def detect_aerial_objects(
    image_path,
    confidence_threshold=CONFIDENCE_THRESHOLD
):

    image = Image.open(
        image_path
    ).convert("RGB")

    image_tensor = to_tensor(image).to(device)

    model.eval()

    with torch.no_grad():
        prediction = model([image_tensor])[0]

    boxes = prediction["boxes"].cpu()
    labels = prediction["labels"].cpu()
    scores = prediction["scores"].cpu()

    keep = scores >= confidence_threshold

    boxes = boxes[keep]
    labels = labels[keep]
    scores = scores[keep]

    result_image = image.copy()
    draw = ImageDraw.Draw(result_image)

    detections = []

    for box, label, score in zip(
        boxes,
        labels,
        scores
    ):

        x1, y1, x2, y2 = box.tolist()

        label_id = int(label)

        if label_id < 1 or label_id > len(class_names):
            continue

        class_name = class_names[label_id - 1]
        confidence = float(score)

        draw.rectangle(
            [x1, y1, x2, y2],
            outline="red",
            width=2
        )

        text = f"{class_name}: {confidence:.2f}"

        text_y = max(0, y1 - 18)

        text_box = draw.textbbox(
            (x1, text_y),
            text
        )

        draw.rectangle(
            text_box,
            fill="red"
        )

        draw.text(
            (x1, text_y),
            text,
            fill="white"
        )

        detections.append({
            "class": class_name,
            "confidence": round(confidence, 4),
            "bbox_xyxy": [
                round(x1, 2),
                round(y1, 2),
                round(x2, 2),
                round(y2, 2)
            ]
        })

    return result_image, detections


# ================================================================
# 18. TEST ON VALIDATION IMAGES
# ================================================================

test_records = [
    record for record in val_records
    if len(record["boxes"]) > 0
]

if not test_records:
    test_records = val_records

NUM_TEST_IMAGES = min(4, len(test_records))

for record in test_records[:NUM_TEST_IMAGES]:

    result_image, detections = detect_aerial_objects(
        record["image_path"],
        confidence_threshold=CONFIDENCE_THRESHOLD
    )

    plt.figure(figsize=(12, 10))
    plt.imshow(result_image)
    plt.axis("off")
    plt.title(
        f"Aerial Object Detection | "
        f"Predictions: {len(detections)}"
    )
    plt.show()

    for detection in detections:
        print(detection)


# ================================================================
# 19. SAVE FINAL MODEL
# ================================================================

torch.save(
    {
        "model_state_dict": model.state_dict(),
        "class_names": class_names,
        "num_classes": num_classes,
        "history": history
    },
    CHECKPOINT_PATH
)

print("\nFinal model saved:", CHECKPOINT_PATH)


# ================================================================
# 20. DOWNLOAD TRAINED MODEL
# ================================================================

files.download(CHECKPOINT_PATH)


# ================================================================
# 21. OPTIONAL: UPLOAD A NEW IMAGE FOR DETECTION
# ================================================================

print("\nOptionally upload a new aerial image for inference.")

new_images = files.upload()

for filename in new_images:

    if Path(filename).suffix.lower() not in IMAGE_EXTENSIONS:
        print("Skipping non-image file:", filename)
        continue

    result_image, detections = detect_aerial_objects(
        filename,
        confidence_threshold=CONFIDENCE_THRESHOLD
    )

    plt.figure(figsize=(12, 10))
    plt.imshow(result_image)
    plt.axis("off")
    plt.title("Faster R-CNN Aerial Object Detection")
    plt.show()

    print("\nDetections for:", filename)

    for detection in detections:
        print(detection)

print("\nALL STEPS COMPLETED.")
```
