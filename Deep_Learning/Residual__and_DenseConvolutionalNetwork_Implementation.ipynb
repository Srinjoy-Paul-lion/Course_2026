```python
# ============================================================
# RESNET-STYLE CNN ON MNIST
# ============================================================

import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import layers, Model
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------

EPOCHS = 10
BATCH_SIZE = 128
DROPOUT_RATE = 0.20

# ------------------------------------------------------------
# Load MNIST
# ------------------------------------------------------------

(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Add channel dimension
x_train = x_train[..., tf.newaxis]
x_test = x_test[..., tf.newaxis]

# Normalize
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# One-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print("Training data:", x_train.shape)
print("Testing data :", x_test.shape)


# ------------------------------------------------------------
# Residual Block
# ------------------------------------------------------------

def residual_block(x, filters, dropout_rate=0.2):

    shortcut = x

    # First convolution
    y = layers.Conv2D(
        filters,
        kernel_size=3,
        padding="same",
        activation="relu"
    )(x)

    y = layers.BatchNormalization()(y)

    # Second convolution
    y = layers.Conv2D(
        filters,
        kernel_size=3,
        padding="same"
    )(y)

    y = layers.BatchNormalization()(y)

    # Match dimensions for shortcut
    if x.shape[-1] != filters:
        shortcut = layers.Conv2D(
            filters,
            kernel_size=1,
            padding="same"
        )(shortcut)

        shortcut = layers.BatchNormalization()(shortcut)

    # Residual connection
    y = layers.Add()([shortcut, y])

    # Activation
    y = layers.Activation("relu")(y)

    # Dropout
    y = layers.Dropout(dropout_rate)(y)

    return y


# ------------------------------------------------------------
# Build ResNet
# ------------------------------------------------------------

inputs = layers.Input(shape=(28, 28, 1))

# Initial convolution
x = layers.Conv2D(
    32,
    kernel_size=3,
    padding="same",
    activation="relu"
)(inputs)

x = layers.BatchNormalization()(x)


# ---------------- Residual Block 1 ----------------

x = residual_block(
    x,
    filters=32,
    dropout_rate=DROPOUT_RATE
)

x = layers.MaxPooling2D(pool_size=2)(x)


# ---------------- Residual Block 2 ----------------

x = residual_block(
    x,
    filters=64,
    dropout_rate=DROPOUT_RATE
)

x = layers.MaxPooling2D(pool_size=2)(x)


# ---------------- Residual Block 3 ----------------

x = residual_block(
    x,
    filters=128,
    dropout_rate=DROPOUT_RATE
)

x = layers.MaxPooling2D(pool_size=2)(x)


# ---------------- Residual Block 4 ----------------

x = residual_block(
    x,
    filters=256,
    dropout_rate=DROPOUT_RATE
)


# ------------------------------------------------------------
# Classification
# ------------------------------------------------------------

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dropout(DROPOUT_RATE)(x)

outputs = layers.Dense(
    10,
    activation="softmax"
)(x)


# ------------------------------------------------------------
# Create Model
# ------------------------------------------------------------

model = Model(inputs, outputs)

model.summary()


# ------------------------------------------------------------
# Compile
# ------------------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

history = model.fit(
    x_train,
    y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.10,
    verbose=1
)


# ------------------------------------------------------------
# Test
# ------------------------------------------------------------

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nTest Loss:", test_loss)
print("Test Accuracy:", test_accuracy)


# ------------------------------------------------------------
# Plot Accuracy
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("ResNet on MNIST")

plt.legend()
plt.grid(True)

plt.show()


```

```python
# ============================================================
# DENSENET-STYLE CNN ON MNIST
# ============================================================

import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import layers, Model
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------

EPOCHS = 10
BATCH_SIZE = 128

DROPOUT_RATE = 0.20
GROWTH_RATE = 16


# ------------------------------------------------------------
# Load MNIST
# ------------------------------------------------------------

(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train[..., tf.newaxis]
x_test = x_test[..., tf.newaxis]

# Normalize
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# One-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print("Training data:", x_train.shape)
print("Testing data :", x_test.shape)


# ------------------------------------------------------------
# Dense Layer
# ------------------------------------------------------------

def dense_layer(x):

    # Batch Normalization
    y = layers.BatchNormalization()(x)

    # ReLU
    y = layers.Activation("relu")(y)

    # Convolution
    y = layers.Conv2D(
        GROWTH_RATE,
        kernel_size=3,
        padding="same"
    )(y)

    # Dropout
    y = layers.Dropout(
        DROPOUT_RATE
    )(y)

    # Dense connection
    y = layers.Concatenate()([x, y])

    return y


# ------------------------------------------------------------
# Dense Block
# ------------------------------------------------------------

def dense_block(x, number_of_layers):

    for i in range(number_of_layers):

        x = dense_layer(x)

    return x


# ------------------------------------------------------------
# Transition Block
# ------------------------------------------------------------

def transition_block(x):

    # Batch Normalization
    x = layers.BatchNormalization()(x)

    # ReLU
    x = layers.Activation("relu")(x)

    # 1x1 convolution
    filters = int(x.shape[-1] * 0.5)

    x = layers.Conv2D(
        filters,
        kernel_size=1,
        padding="same",
        activation="relu"
    )(x)

    x = layers.Dropout(
        DROPOUT_RATE
    )(x)

    # Max pooling
    x = layers.MaxPooling2D(
        pool_size=2
    )(x)

    return x


# ------------------------------------------------------------
# Build DenseNet
# ------------------------------------------------------------

inputs = layers.Input(
    shape=(28, 28, 1)
)

# Initial convolution
x = layers.Conv2D(
    32,
    kernel_size=3,
    padding="same",
    activation="relu"
)(inputs)


# ================= Dense Block 1 =================

x = dense_block(
    x,
    number_of_layers=2
)

x = transition_block(x)


# ================= Dense Block 2 =================

x = dense_block(
    x,
    number_of_layers=2
)

x = transition_block(x)


# ================= Dense Block 3 =================

x = dense_block(
    x,
    number_of_layers=2
)

x = transition_block(x)


# ================= Dense Block 4 =================

x = dense_block(
    x,
    number_of_layers=2
)


# ------------------------------------------------------------
# Classification
# ------------------------------------------------------------

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dropout(
    DROPOUT_RATE
)(x)

outputs = layers.Dense(
    10,
    activation="softmax"
)(x)


# ------------------------------------------------------------
# Create Model
# ------------------------------------------------------------

model = Model(
    inputs,
    outputs
)

model.summary()


# ------------------------------------------------------------
# Compile
# ------------------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

history = model.fit(
    x_train,
    y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.10,
    verbose=1
)


# ------------------------------------------------------------
# Test
# ------------------------------------------------------------

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nTest Loss:", test_loss)
print("Test Accuracy:", test_accuracy)


# ------------------------------------------------------------
# Plot Accuracy
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("DenseNet on MNIST")

plt.legend()
plt.grid(True)

plt.show()

```
