### Answers to the Questions

1. **Why does the model assign different confidence scores to different detections?**
   
   **Ans:-** The confidence score represents how certain the neural network's classifier head is that a bounding box contains an object of a specific category. Variations occur due to factors like object occlusion, lighting changes, orientation variations, scale differences, and how closely the object matches the features learned during training from the COCO dataset.

2. **What happens when the confidence threshold increases?**
   
   **Ans:-** Increasing the threshold reduces the number of detections (increases precision but lowers recall). Only detections where the model is highly certain are kept, filtering out low-confidence predictions (which are often background noise or false positives).

3. **Can the pre-trained model detect an object category that it was not trained to recognize?**

   **Ans:-** No, a standard pre-trained model cannot correctly identify or label an entirely unseen category since its final classification layer is strictly mapped to the 80/91 COCO categories. It will either completely ignore the unknown object or misclassify it as the closest matching known class.
